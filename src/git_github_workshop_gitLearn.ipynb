## 📌 مفاهیم پایه و دستورات اولیه (جلسه اول)

### سیستم‌های کنترل ورژن (VCS)

- **VCS (Version Control System):** ابزاری برای مدیریت، پیگیری و ذخیره‌سازی تغییرات سورس‌کد در طول زمان که امکان بازگشت به نسخه‌های قبلی و همکاری هم‌زمان چند برنامه‌نویس را فراهم می‌کند.
- **تفاوت Git و GitHub:**
  - **Git:** ابزار کنترل ورژن توزیع‌شده (Distributed VCS) که روی سیستم محلی (Local) خودمان نصب و اجرا می‌شود و نیازی به اینترنت ندارد.
  - **GitHub:** سرویس ابری میزبانی مخازن Git برای اشتراک‌گذاری کد، همکاری گروهی، بررسی کد (Code Review) و مدیریت پروژه.

### معماری سه‌گانه گیت (Three States)

برای درک عمیق کارکرد گیت، باید سه فضای اصلی آن را بشناسیم:
1. **Working Directory (فضای کاری):** پوشه‌ای که فایل‌های پروژه در آن قرار دارند و در حال ویرایش آن‌ها هستید.
2. **Staging Area / Index (ناحیه آماده‌سازی):** مثل «دکه‌ی بسته‌بندی اداره پست»؛ فایل‌هایی که تغییر داده‌اید را قبل از ثبت نهایی در این فضا می‌چینید تا مشخص شود دقیقاً چه چیزهایی قرار است در بسته‌ی بعدی ثبت شوند.
3. **Repository / `.git` (مخزن اصلی):** بایگانی دائمی تمام Snapshotها (کامیت‌ها) که اشاره‌گر `HEAD` به آخرین نسخه‌ی فعلی آن اشاره می‌کند.

### ثبت تغییرات و کنترل شدن فایل‌ها

  - `git commit -m "Message"` ← ثبت کامیت همراه با پیام توضیحی کوتاه و گویا.
  - `git commit -a -m "Message"` (یا `-am`) ← میانبری که فایل‌های **Tracked** و تغییرکرده را هم‌زمان Stage و Commit می‌کند.

🔹 **نکته‌ی مهم:** میانبر `-am` فایل‌های جدید (`Untracked`) را شامل نمی‌شود و برای آن‌ها حتماً باید ابتدا `git add` بزنید.

</div>

### history همیشه مهمه

- `git log` ← مشاهده تاریخچه‌ی کامیت‌ها به همراه نویسنده، تاریخ و پیام.
  - `git log --oneline --graph --all` ← نمایش فشرده و گرافیکی تاریخچه.
  - **Hash:** شناسه‌ی یکتای هر کامیت؛ یک کد ۴۰ کاراکتری SHA-1 (که معمولاً ۷ کاراکتر اول آن نمایش داده می‌شود).
- `git show <hash>` ← نمایش جزئیات کامل و تغییرات (Diff) یک کامیت مشخص.
- `git diff` ← مقایسه‌ی تغییرات **Working Directory** با **Staging Area** (تغییراتی که هنوز `add` نشده‌اند).
- `git diff --staged` ← مقایسه‌ی تغییرات **Staged** با آخرین کامیت (`HEAD`).

</div>

````python
%%bash
cd gitLearn

# ۱. مشاهده تاریخچه‌ی واقعی مخزن به صورت خلاصه و گرافیکی (۱۵ کامیت آخر)
git log --oneline --graph --all -n 15

# ۲. مشاهده جزئیات یک کامیت مشخص با Hash (اینجا: کامیتی که با --amend اصلاح شده بود)
git show 1a10c68 --stat

# ۳. مقایسه‌ی دو کامیت قدیمی با هم
git diff bf804c1 1a10c68 --stat

# ۴. ایجاد تغییر جدید در فایل بدون زدن git add
printf '\n# calculator entry point\n' >> calucator.py

# ۵. مشاهده خطوط اضافه یا کم شده نسبت به Staging Area
git diff

# ۶. بردن تغییر به استیج و مقایسه با آخرین کامیت
git add calucator.py
git diff --staged
````

- `git mv <old> <new>` ← تغییر نام یا جابه‌جایی فایل و Stage کردن تغییر.
- **فایل `.gitignore`:** برای نادیده گرفتن فایل‌های موقت، لاگ‌ها (`*.log`)، تاریخچه محاسبات (`history.txt`)، کش پایتون (`__pycache__/`)، تنظیمات IDE یا فایل‌های محرمانه مانند `.env` استفاده می‌شود.

</div>

### Ctrl + Z (لغو تغییرات محلی)

- `git checkout -- <file>` ← روش قدیمی برای دور انداختن تغییرات محلیِ ثبت‌نشده در فایل و برگرداندن آن به وضعیت آخرین کامیت.
- `git restore <file>` ← دستور جدیدتر و خواناتر برای بازگرداندن فایل در Working Directory به وضعیت آخرین کامیت.
- **هشدار:** این دستورها تغییرات ذخیره‌نشده را برای همیشه پاک می‌کنند.

### بیا پایین از استیج

- `git restore --staged <file>` ← خارج کردن فایل از Staging Area و برگرداندن آن به Working Directory، بدون حذف تغییرات داخل فایل.

### برگشت خوردن کامیت

- `git commit --amend` ← اصلاح آخرین کامیت؛ مثلاً تغییر پیام کامیت یا اضافه کردن فایل فراموش‌شده به آخرین کامیت. **نکته:** این دستور تاریخچه را بازنویسی (Rewrite) می‌کند و نباید روی کامیت‌هایی که Push شده‌اند استفاده شود.
- `git revert <commit>` ← ساخت یک کامیت جدید که اثر کامیت مشخص‌شده را خنثی می‌کند؛ برای تاریخچه‌ی مشترک و آنلاین کاملاً ایمن است.

### رو Save کار کردن (Stash)

- `git stash` ← کنار گذاشتن موقت تغییرات Commit‌نشده در یک حافظه‌ی پنهان برای تمیز شدن Working Directory.
- `git stash pop` ← برگرداندن آخرین تغییرات Stash به پروژه و حذف آن مورد از فهرست Stash.
- `git stash apply` ← برگرداندن تغییرات بدون حذف آن مورد از فهرست Stash.

</div>

````python
%%bash
cd gitLearn

# ۱. تست git restore (خراب کردن عمدی فایل و بازگردانی آن)
printf '\nBROKEN SYNTAX ###\n' >> operations.py
git status -s
git restore operations.py
git status -s

# ۲. تست git restore --staged (خارج کردن فایل از استیج)
printf '\n# Main CLI\n' >> calucator.py
git add calucator.py
git status -s
git restore --staged calucator.py
git status -s

# ۳. تست git commit --amend (اصلاح کامیت آخر با اضافه کردن calucator.py)
git add calucator.py
git commit --amend -m "chore: ignore log files and comment calucator.py"

# ۴. تست git stash و git stash pop (ذخیره موقت کار نیمه‌تمام)
printf '\n# WIP: calculator menu\n' >> calucator.py
git stash push -m "wip: calculator menu"
git status -s  # محیط کاملاً تمیز شد
git stash pop  # بازگرداندن کار نیمه‌تمام
git commit -am "feat: add WIP note to calucator.py"

# ۵. نمونه‌ی واقعی git revert در تاریخچه‌ی همین پروژه (e33fe7d کامیت 9c55da5 را خنثی کرد)
git show e33fe7d --stat

# ۶. ایجاد باگ عمدی و خنثی کردن ایمن آن با کامیت جدید
printf "\nraise RuntimeError('Bug!')\n" >> operations.py
git commit -am "bug: accidental runtime error"
git revert HEAD --no-edit
git log --oneline -n 4
````

### شاخه (Branch)

**مفهوم Branch:** شاخه راهی برای توسعه‌ی مستقل یک قابلیت (Feature) یا اصلاح باگ (Bugfix)، بدون دست‌کاری مستقیم و ناامن کردن شاخه‌ی اصلی (`main`) است. (مثل اضافه کردن دسته به یک لیوان بدون خراب کردن بدنه اصلی لیوان).

- `git branch` ← مشاهده لیست شاخه‌ها و تشخیص شاخه‌ی فعلی (علامت `*`).
- `git branch <branchName>` ← ساخت شاخه‌ی جدید.
- `git checkout -b <branchName>` ← ساخت شاخه و جابه‌جایی به آن به‌صورت هم‌زمان.
- `git switch -c <branchName>` ← دستور مدرن و استاندارد برای ساخت شاخه و رفتن روی آن.

### از این شاخه به اون شاخه پریدن

- `git checkout <branchName>` یا `git switch <branchName>` ← جابه‌جایی بین شاخه‌ها (اشاره‌گر `HEAD` به شاخه‌ی مقصد منتقل می‌شود).

### ادغام‌سازی (Merge)

- `git merge <branchName>` ← ادغام تغییرات شاخه‌ی مشخص‌شده در شاخه‌ی فعلی. ابتدا باید به شاخه‌ی مقصد (مثلاً `main`) برویم و سپس دستور ادغام را بزنیم.

##### روش‌های بردن Branch روی Main
1. **Fast-Forward (`ff`):** وقتی شاخه‌ی مقصد از زمان انشعاب، کامیت جدیدی نداشته باشد؛ اشاره‌گر `main` به سادگی جلو می‌رود.
2. **3-Way Merge:** وقتی هر دو شاخه از نقطه‌ی مشترک تغییر کرده‌اند؛ گیت یک Merge Commit جدید می‌سازد.
3. **Rebase:** انتقال خطی کامیت‌ها به نوک شاخه‌ی مقصد.

### حذف کردن شاخه

- `git branch -d <branchName>` ← حذف امن شاخه‌ای که تغییراتش قبلاً ادغام شده‌اند.
- `git branch -D <branchName>` ← حذف اجباری شاخه (حتی اگر ادغام نشده باشد)؛ با احتیاط استفاده شود.

</div>

````python
%%bash
cd gitLearn

# ۰. لیست شاخه‌ها: شاخه‌ی محلی main + شاخه‌های ریموت پروژه (usrName/...)
git branch -a

# ۱. ساخت شاخه جدید برای قابلیت فاکتوریل و رفتن روی آن
git switch -c feature/factorial

# ۲. اضافه کردن تابع factorial
cat << 'EOF' >> operations.py

def factorial(n):
    if n < 0:
        raise ValueError("Error: Negative number!")
    return 1 if n in (0, 1) else n * factorial(n - 1)
EOF

git commit -am "feat: add factorial operation"

# ۳. بازگشت به شاخه اصلی (main) و انجام Fast-Forward Merge
git switch main
git merge feature/factorial

# ۴. حذف شاخه ادغام‌شده و مشاهده گراف
git branch -d feature/factorial
git log --oneline --graph -n 4
````

### تضاد (Conflict)

**چه زمانی Conflict رخ می‌دهد؟** وقتی دو شاخه‌ی مختلف، هم‌زمان **یک خط مشخص از یک فایل مشترک** را به دو شکل متفاوت تغییر داده باشند، گیت نمی‌تواند به‌صورت خودکار تصمیم بگیرد کدام تغییر درست است و از برنامه‌نویس کمک می‌خواهد.

**مراحل حل Conflict:**
>1. `git status` را اجرا می‌کنیم تا فایل‌های دارای تضاد (`both modified`) را ببینیم.
>2. فایل‌های متعارض را در ویرایشگر باز می‌کنیم.
>3. بخش‌های علامت‌گذاری‌شده با `<<<<<<< HEAD` (تغییر شاخه فعلی)، `=======` و `>>>>>>> branch-name` (تغییر شاخه ورودی) را بررسی و نسخه‌ی نهایی موردنظر را انتخاب می‌کنیم.
>4. فایل را ذخیره و با `git add <file>` آن را Stage می‌کنیم.
>5. با `git commit` ادغام (3-Way Merge) را نهایی می‌کنیم.

<p style="text-align:center"><b>یا</b></p>

> اگر از انجام Merge منصرف شدیم، تا زمانی که Conflict حل نشده است می‌توانیم با دستور `git merge --abort` همه‌چیز را به وضعیت قبل از Merge برگردانیم.

</div>

````python
%%bash
cd gitLearn

# در تاریخچه‌ی واقعی این پروژه، دو نفر هم‌زمان نام متغیرهای calucator.py را تغییر دادند:
#   b099899 (روی main):          num1 & num2 -> n1 & n2
#   668193f (روی usrName/complateUi): num1 & num2 -> number1 & number2
# همین ادغام را دوباره اجرا می‌کنیم تا Conflict را ببینیم.

# ۱. ساخت شاخه‌ی تمرینی از کامیت b099899
git switch -c conflict-demo b099899

# ۲. تلاش برای ادغام کامیت دیگر (این دستور عمداً با خطای Conflict متوقف می‌شود)
git merge 668193f --no-edit || echo "⚠️ Conflict detected as expected!"

# ۳. مشاهده فایل‌های دارای تضاد و نشانگرهای تضاد داخل فایل
git status -s
grep -n -E '^(<<<<<<<|=======|>>>>>>>)' calucator.py

# ۴. حل Conflict: نسخه‌ی نهایی را از ادغام واقعی (c784d15) برمی‌داریم، Stage و Commit می‌کنیم
git checkout c784d15 -- calucator.py
git add calucator.py
git commit -m "merge: resolve variable names conflict using 3-way merge"

# ۵. مشاهده‌ی گراف و پاک‌سازی شاخه‌ی تمرینی
git log --oneline --graph -n 5
git switch main
git branch -D conflict-demo
````

### Setup و اتصال به GitHub از طریق SSH

برای ارتباط امن و بدون نیاز به وارد کردن رمز عبور در هر بار Push/Pull، از جفت‌کلید SSH استفاده می‌کنیم:

1. ساخت کلید جدید با الگوریتم امن Ed25519:
   ```bash
   ssh-keygen -t ed25519 -C "email@example.com"
   ```
2. مشاهده‌ی فایل‌های کلید ساخته‌شده:
   ```bash
   ls ~/.ssh
   ```
3. نمایش و کپی کردن کلید عمومی (Public Key):
   ```bash
   cat ~/.ssh/id_ed25519.pub
   ```
4. در سایت GitHub وارد **Settings → SSH and GPG keys** شوید.
5. گزینه‌ی **New SSH key** را بزنید، یک عنوان دلخواه بنویسید و کلید عمومی را وارد و ذخیره کنید.
6. تست اتصال به گیت‌هاب:
   ```bash
   ssh -T git@github.com
   ```

### حذف و اضافه مخزن‌ها

- `git clone <ssh_url>` ← دریافت یک کپی کامل از مخزن ریموت (به همراه تمام تاریخچه و برنچ‌ها) روی سیستم محلی.
- `rm -rf <repoName>` ← حذف کامل پوشه‌ی پروژه از سیستم (غیرقابل بازگشت؛ با احتیاط اجرا شود).

### Remote چیه؟ HEAD کجاست؟

- **Remote:** نام مستعار و آدرس مخزن آنلاین (مثلاً روی GitHub) که با مخزن محلی ما در ارتباط است؛ نام پیش‌فرض آن معمولاً `origin` است.
- **HEAD:** اشاره‌گری که نشان می‌دهد هم‌اکنون روی کدام شاخه یا کامیت ایستاده‌ایم.

### اوضاع گیت‌هاب چطوره؟

در مخازن متصل به GitHub، دستور `git status` وضعیت شاخه‌ی محلی را نسبت به شاخه‌ی ریموت (`origin/main`) گزارش می‌دهد:
- `Your branch is up to date with 'origin/main'` ← همگام هستید.
- `Your branch is ahead of 'origin/main' by X commits` ← شما X کامیت جدید دارید که هنوز Push نکرده‌اید.
- `Your branch is behind 'origin/main' by X commits` ← همکارانتان X کامیت جدید روی سرور زده‌اند که شما هنوز Pull نکرده‌اید.

### وضعیت ریموت‌ها

- `git remote add origin <url>` ← اتصال یک مخزن محلی به مخزن خالی در گیت‌هاب.
- `git remote -v` ← نمایش آدرس‌های Fetch و Push ریموت‌ها.
- `git remote show origin` ← نمایش جزئیات کامل وضعیت شاخه‌ها در ریموت `origin`.

### هل دادن فایل‌ها به گیت (Push)

- `git push -u origin main` ← اولین ارسال شاخه `main` به `origin` و تنظیم ارتباط (Upstream) بین آن دو.
- `git push` ← ارسال کامیت‌های محلی به ریموت و شاخه‌ی تنظیم‌شده.
- `git push <remote> <branch>` ← ارسال یک شاخه‌ی خاص به ریموت (مثلاً `git push origin feature/login`).

### نوتیف گرفتن از گیت (Fetch & Pull)

- `git fetch` ← دریافت اطلاعات و کامیت‌های جدید از ریموت بدون دست‌کاری فایل‌های فعلی شما (فقط آپدیت کردن دانش گیت از وضعیت سرور).
- `git merge origin/main` ← ادغام تغییرات دریافت‌شده از ریموت در شاخه‌ی فعلی.
- `git pull` ← ترکیبی از `git fetch` و `git merge` در یک دستور (دریافت و ادغام هم‌زمان).
  - `git pull <remote> <branch>` ← دریافت و ادغام تغییرات یک شاخه‌ی مشخص.

</div>

````python
%%bash
cd gitLearn

# ۱. بررسی وضعیت همگام بودن با ریموت
#    (تمرین‌های بالا چند کامیت محلی ساخته‌اند، پس باید «ahead of origin/main» ببینید)
git status

# ۲. دریافت آخرین وضعیت از ریموت با fetch و pull
git fetch origin
git pull --no-rebase --no-edit origin main

# ۳. ارسال به ریموت (Push)
#    ⚠️ مخزن Shahed-CE-Association/gitLearn مخزن مشترک است و شما معمولاً دسترسی Push ندارید،
#    پس این دستورها فقط برای یادگیری کامنت شده‌اند:
# git push -u origin main
# git push origin usrName/feature-name
````

### 3-Way Merge در برابر Rebase

وقتی روی شاخه‌ی `feature` کار می‌کنید و هم‌زمان همکارانتان روی `main` کامیت‌های جدیدی ثبت کرده‌اند، برای آوردن آخرین تغییرات `main` به شاخه‌ی خود دو راه دارید:
- **3-Way Merge (`git merge main`):** تاریخچه واقعی را حفظ می‌کند و یک گره‌ی جدید (Merge Commit) می‌سازد.
- **Rebase (`git rebase main`):** پایه‌ی شاخه‌ی شما را جابه‌جا کرده و کامیت‌های شما را دانه به دانه با **Hash جدید** روی آخرین کامیت `main` قرار می‌دهد تا تاریخچه‌ای کاملاً خطی و تمیز بسازد.

⚠️ **قانون طلایی Rebase:** هرگز روی شاخه‌های عمومی و مشترک (مثل `main`) که دیگران روی آن کار می‌کنند Rebase نزنید، چون تاریخچه را بازنویسی می‌کند.

### Conflict در GitHub و هنگام Rebase

هنگام بروز Conflict در Rebase:
1. با `git status` فایل دارای تضاد را پیدا می‌کنیم.
2. فایل را ویرایش و نشانگرهای Conflict را پاک می‌کنیم.
3. فایل اصلاح‌شده را با `git add <file>` استیج می‌کنیم.
4. به جای `git commit`، دستور `git rebase --continue` را اجرا می‌کنیم (یا برای لغو کامل، `git rebase --abort`).

### آقا هل نده! (Force Push)

پس از Rebase یا Amend روی شاخه‌ای که قبلاً Push شده، گیت اجازه Push معمولی نمی‌دهد چون تاریخچه تغییر کرده است:
- `git push --force` یا `git push -f` ← بازنویسی اجباری و خطرناک تاریخچه‌ی ریموت (ممکن است کامیت‌های همکارانتان را پاک کند).
- `git push --force-with-lease` ← **روش امن و حرفه‌ای**؛ ابتدا چک می‌کند که کسی از آخرین باری که شما Fetch کرده‌اید کامیتی روی سرور نزده باشد و سپس Push را انجام می‌دهد.

</div>

````python
%%bash
cd gitLearn

# نمونه‌ی واقعی Rebase در این پروژه: کامیت‌های 3e4dd0f و 7426547 ("Now we wanna do rebase" / "we did rebase.")
git log --oneline --graph -n 8

# ۱. ساخت شاخه feature از یک نقطه‌ی قدیمی‌تر main (کامیت 2d78e5a) و اضافه کردن یک فایل
git switch -c feature/rebase-demo 2d78e5a
echo "rebase demo" > rebase_demo.txt
git add rebase_demo.txt
git commit -m "feat: add rebase demo file"
echo "Hash قبل از Rebase: $(git rev-parse --short HEAD)"

# ۲. ری‌بیس کردن شاخه روی آخرین نقطه main (کامیت‌ها Hash جدید می‌گیرند)
git rebase main
echo "Hash بعد از Rebase: $(git rev-parse --short HEAD)"

# ۳. ادغام خطی (Fast-Forward) در main و مشاهده تاریخچه کاملاً خطی
git switch main
git merge feature/rebase-demo
git branch -d feature/rebase-demo

git log --oneline --graph -n 6

# ۴. اگر این شاخه قبلاً Push شده بود، بعد از Rebase باید با --force-with-lease ارسال می‌شد:
# git push --force-with-lease origin usrName/feature-name
````

### چنگال (Fork)

- **Fork:** ساخت یک کپی مستقل از مخزن شخص دیگر در حساب GitHub خودمان، برای مشارکت در پروژه‌های متن‌باز (Open Source) یا مخازنی که دسترسی مستقیم Push به آن‌ها نداریم.
- **Upstream:** نامی که به‌صورت قراردادی به مخزن اصلی (مخزنی که از روی آن Fork زده‌اید) می‌دهند.
- **Sync Fork:** به‌روز نگه داشتن مخزن Fork‌شده‌ی خودتان با آخرین تغییرات مخزن اصلی (`upstream`).

### وقتی فایل‌ها پشت در می‌مونن (Pull Request / Merge Request)

- **Pull Request (PR) / Merge Request (MR):** درخواستی رسمی در GitHub یا GitLab که از مالک مخزن می‌خواهید تغییرات شاخه‌ی شما را بررسی (Code Review) کرده و در صورت تأیید در شاخه‌ی اصلی (`main`) ادغام کند.

#### گردش کار استاندارد مشارکت در پروژه‌ها (Fork & PR Workflow):
1. مخزن اصلی را **Fork** کنید و نسخه‌ی خودتان را **Clone** نمایید (`git clone ...`).
2. یک شاخه‌ی جدید بسازید (`git switch -c feature/my-feature`).
3. تغییرات را اعمال و کامیت کنید (`git commit -am "..."`).
4. شاخه را به مخزن خودتان در گیت‌هاب پوش کنید (`git push -u origin feature/my-feature`).
5. در صفحه GitHub دکمه‌ی **Compare & pull request** را بزنید، توضیحات تغییرات را بنویسید و **Create pull request** را ثبت کنید.
